```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('IBM-HR-Employee-Attrition.csv')
```

```python
pd.set_option('display.max_columns', None)
```

```python
df.head()
```

```python
df.columns
```

```python
df
```

```python
cols_to_drop = [col for col in df.columns if df[col].nunique() == 1]
cols_to_drop
```

```python
df.drop(cols_to_drop, axis = 1, inplace = True)
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df.head()
```

```python
df['Attrition'] = df['Attrition'].map({'Yes': 1, 'No' : 0})
df['OverTime'] = df['OverTime'].map({'Yes': 1, 'No' : 0})
```

```python
df.nunique().sort_values()
```

```python
cat_cols = ['BusinessTravel', 'Department', 'Education', 'EducationField',
            'EnvironmentSatisfaction', 'Gender', 'JobInvolvement', 'JobLevel',
            'JobRole', 'JobSatisfaction', 'MaritalStatus', 'OverTime',
            'PerformanceRating', 'RelationshipSatisfaction', 
            'StockOptionLevel', 'WorkLifeBalance']

for col in cat_cols:
    df[col] = df[col].astype('category')
```

```python
df.info()
```

```python
df['JobSatisfaction'] = df['JobSatisfaction'].map({
    1: 'Low',
    2: 'Medium',
    3: 'High',
    4: 'Very High'
})
```

```python
df['Attrition'].value_counts(normalize= True) * 100
```

### Which department has the highest attrition rate, and how significant is the difference?

```python
plt.figure(figsize = (10,5))
sns.barplot(df, x = 'Department', y = 'Attrition')
plt.show()
```

```python
df['Department'].value_counts()
```

```python
df.groupby('Department')['Attrition'].mean() * 100
```

##### `Sales` department has the highest attrition rate at 20.6%, nearly 7% higher than `R&D`. `HR's` high rate (19%) should be interpreted cautiously due to small sample size (n=63).

### Do employees who work overtime have a significantly higher attrition rate compared to those who don't?

```python
plt.figure(figsize = (8,5))
sns.barplot(df, x = 'OverTime', y = 'Attrition')
plt.show()
```

```python
df.groupby('OverTime')['Attrition'].mean() * 100
```

##### Employees working overtime are 3x more likely to resign (30.5%) compared to those who don't (10.4%). Overtime is the strongest attrition driver identified so far.

### Does job satisfaction level impact employee attrition rate?

```python
plt.figure(figsize = (10,5))
sns.barplot(df, x = 'JobSatisfaction', y = 'Attrition')
plt.show()
```

```python
df.groupby('JobSatisfaction')['Attrition'].mean()*100
```

##### Employees with the lowest job satisfaction (level 1) have twice the attrition rate (22.8%) compared to highly satisfied employees (11.3%). Improving job satisfaction could significantly reduce turnover.

### Do younger employees resign more than older ones?

```python
plt.figure(figsize=(12,5))
sns.histplot(df, x = 'Age', hue = 'Attrition', kde = True, bins= 30)
plt.show()
```

```python
df[df['Age'] <= 20]['Attrition'].mean() * 100
```

```python
df[df['Age'] <= 20]['Attrition'].count()
```

##### Employees aged 25-30 show the highest attrition. While under-20 age group shows 57% attrition rate, this is based on only 28 employees and should be interpreted with caution.

### Do unmarried employees have a higher attrition rate than married employees?

```python
plt.figure(figsize = (10,5))
sns.barplot(df, x = 'MaritalStatus', y = 'Attrition')
plt.show()
```

```python
df.groupby('MaritalStatus')['Attrition'].mean()* 100
```

##### Single employees have the highest attrition rate (25.5%), more than double that of married employees (12.5%). This may suggest that single employees have fewer financial obligations and are more willing to take career risks by switching jobs.

### Do employees who changed companies more times have higher attrition?

```python
sns.barplot(df, x = 'NumCompaniesWorked', y = 'Attrition')
```

```python
df.groupby('NumCompaniesWorked')['Attrition'].mean()*100
```

```python
df['NumCompaniesWorked'].value_counts()
```

##### Employees who have worked at 5-7 companies show the highest attrition (22-25%), suggesting job-hopping history is a strong attrition predictor. However, data for 8-9 companies should be interpreted cautiously due to small sample sizes (n<55).

### Do employees who do not get promotions have higher attrition rates?


```python
plt.figure(figsize = (10,5))
sns.barplot(df, x = 'YearsSinceLastPromotion', y = 'Attrition')
plt.show()
```

```python
df.groupby('YearsSinceLastPromotion')['Attrition'].mean()*100
```

```python
df['YearsSinceLastPromotion'].value_counts().sort_index()
```

##### Employees who were never promoted (0 years) and those waiting 7+ years show the highest attrition. However, no clear linear trend exists — suggesting promotion alone is not the primary driver of attrition."

### BusinessTravel effect on Attrition?

```python
plt.figure(figsize = (10,3))
sns.barplot(df, x = 'BusinessTravel', y = 'Attrition')
plt.show()
```

```python
df.groupby('BusinessTravel')['Attrition'].mean()* 100
```

```python
df['BusinessTravel'].value_counts().sort_index()
```

##### Frequent travelers have the highest attrition rate (24.9%), nearly 3x higher than non-travelers (8%). Excessive business travel appears to be a significant factor in employee burnout and resignation.

### Over time  and business travel combinely effect on Attrition?

```python
df.groupby(['OverTime','BusinessTravel'])['Attrition'].mean()*100
```

```python
plt.figure(figsize = (12,4))
sns.barplot(df, x = 'BusinessTravel', y = 'Attrition', hue = 'OverTime')
plt.show()
```

##### The combination of overtime and frequent travel is the most toxic work condition — driving attrition to 41.9%. Employees with neither overtime nor travel show only 4.3% attrition — a 10x difference. Reducing overtime for frequent travelers should be the company's top priority.

### Find the Attrition effect on Income

```python
plt.figure(figsize = (10,5))
sns.histplot(df, x = 'MonthlyIncome', hue = 'Attrition', kde = True, bins = 30)
plt.show()
```

```python
df.groupby('Attrition')['MonthlyIncome'].mean()
```

#####  Employees who resigned earned on average `$2,045` less per month (30% lower) than those who stayed. Low compensation is a significant driver of attrition, particularly for employees earning below `$5,000`/month.
